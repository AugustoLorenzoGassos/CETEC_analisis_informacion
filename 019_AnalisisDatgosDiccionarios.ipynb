#Diccionarios
Los diccionarios son colecciones ordenadas que almacenan y acceden a datos mediante pares de dos puntos key( value:).

Los diccionarios de Python utilizan llaves { }con pares separados por comas key:value

```python
dictionary ={
    key1 : value1,
    key2: value2,
    key3: value3
}
```

```python
laptop= {
    "marca": "Apple",
    "modelo": "Macbook Pro",
    "tamaño": 14,
    "año" : 2023
}
```

```python
print(laptop["modelo"])
```

```python
student_info ={ "name": "Fernando", "age": 9, "grado" : "A"}
student_info["age"]=10
print(student_info["age"])
```

```python
# Dictionary methods
print('Keys:', student_info.keys())
print('Values:', student_info.values())
print('Items:', student_info.items())
```

```python
coleccion_met = {
    "MET-102": {
        "titulo": "Mujer con sombrilla",
        "artista": "Claude Monet",
        "año": 1875,
        "tecnica": ["Óleo", "Lienzo"],
        "expuesto": True },
    "MET-505": {
        "titulo": "La noche estrellada",
        "artista": "Vincent van Gogh",
        "año": 1889,
        "tecnica": ["Óleo", "Lienzo"],
        "expuesto": False },
    "MET-901": {
        "titulo": "Autorretrato con sombrero de paja",
        "artista": "Vincent van Gogh",
        "año": 1887,
        "tecnica": ["Óleo", "Soporte de cartón"],
        "expuesto": True }
    }

obras_expuestas = 0

print("Obras de Vincent van Gogh")
for item_dict in coleccion_met.keys():
  if coleccion_met[item_dict]["artista"] == "Vincent van Gogh":
    print(coleccion_met[item_dict]["titulo"])
  if coleccion_met[item_dict]["expuesto"] == True:
    obras_expuestas += 1

print(f"\nTotal de obras expuestas: {obras_expuestas}")

print(f"\nLa técnica utilizada en la obra {coleccion_met['MET-102']['titulo']} es {coleccion_met['MET-102']['tecnica'][0]}")

```
